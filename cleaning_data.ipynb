```python
import pandas as pd
```

```python
def clean_file(file_name, output_file_name):
    df = pd.read_csv(file_name)
    df[['last_name', 'first_name']] = df['last_name, first_name'].str.split(',', n=1, expand=True)
    df.drop(columns=['last_name, first_name'], inplace=True)
    df.rename(columns={'double': 'doubles'}, inplace=True)
    df.to_csv(output_file_name, index=False)
    
clean_file('hitter_stats.csv', 'cleaned_hitter_stats.csv')
clean_file('hitter_metrics.csv', 'cleaned_hitter_metrics.csv')
clean_file('pitcher_metrics.csv', 'cleaned_pitcher_metrics.csv')  
clean_file('pitcher_stats.csv', 'cleaned_pitcher_stats.csv')
```
